```python
from sympy import Function, Symbol, Eq, symbols, solve
import sympy
```

基本

```python
# 変数
x = sympy.Symbol("x")
y = sympy.Symbol("y")
print(type(x))
```

```python
# 式
expr = x**2 + y + 1
expr
```

```python
# 代入 x=1
display(expr.subs(x, 1))
# 代入 x=y
display(expr.subs(x, y))
```

```python
expr = (x + 1) ** 2
display(expr)
# 式を展開
expr_ex = sympy.expand(expr)
display(expr_ex)
```

```python
# 因数分解
expr_factor = sympy.factor(expr_ex)
display(expr_factor)

display(sympy.factor(x**3 - x**2 - 3 * x + 3))
display(sympy.factor(x * y + x + y + 1))
```

```python
# 方程式を解く
expr = x**2 - 3 * x + 2
display(expr)
display(sympy.solve(expr))

expr = x**2 + x + 1
display(expr)
display(sympy.solve(expr))
```

```python
expr = x + y**2 - 4
# x で解く
display(sympy.solve(expr, x))
# ｙ で解く
display(sympy.solve(expr, y))
```

```python
# 連立方程式
expr1 = 3 * x + 5 * y - 29
expr2 = x + y - 7
display(expr1)
display(expr2)
display(sympy.solve([expr1, expr2]))
```

```python
# 微分
expr = x**3 + 2 * x**2 + x
display(expr)
display(sympy.diff(expr))
display(sympy.Derivative.diff(expr))
display(sympy.Derivative(expr))
```

```python
# 偏微分
expr = x**3 + y**2 - y
# xで偏微分
display(sympy.diff(expr, x))
# yで偏微分
display(sympy.diff(expr, y))
```

```python
# 積分
expr = 3 * x**2 + 4 * x + 1
display(expr)
display(sympy.integrate(expr))
```

```python
# 三角関数・指数関数・対数関数の微分と積分
display(sympy.diff(sympy.cos(x)))
display(sympy.diff(sympy.exp(x)))
display(sympy.diff(sympy.log(x)))
display(sympy.integrate(sympy.cos(x)))
display(sympy.integrate(sympy.exp(x)))
display(sympy.integrate(sympy.log(x)))
```

```python
# インポートの使い方
from sympy import sin, exp, Symbol, diff

x = Symbol("x")
display(diff(sin(x)))
display(diff(exp(x)))
```

```python
# 方程式を解く
from sympy import symbols, solve

# 変数定義
a, b, c, x = symbols("a b c x")
display(a + b + c + x)
# 式を定義
expr = a * x**2 + b * x + c
display(expr)
# x について解く
sol = solve(expr, x)
display(sol)
```

```python
from sympy import expand

a, b, c, d = symbols("a b c d")
expr1 = (a**2 + b**2) * (c**2 + d**2)
display(expr1)
expr1_ex = expand(expr1)
display(expr1_ex)
expr2 = (a * c - b * d) ** 2 + (a * d + b * c) ** 2
display(expr2)
expr2_ex = expand(expr2)
display(expr2_ex)
expr1_ex == expr2_ex
```

```python
from sympy import Matrix, linsolve

x, y, z = symbols("x, y, z")
M = Matrix(
    [
        [1, 2, 3],
        [0, 1, 2],
        [0, 0, 1],
    ]
)
u = Matrix([[1, 2, 3]]).T
system = M, u
linsolve(system, x, y, z)
```

```python
from sympy import symbols, plot

x = symbols("x")
# 数式は英語で numerial expression または numerical formula
expr = x**2 - 12 * x + 8
display(expr)
plot(expr, (x, -20, 20))
```

```python
from sympy import symbols, plot, Eq
import sympy

x = symbols("x")
a, b, c, d, e = symbols("a, b, c, d, e")
expr = x**2 - 12 * x + 8
display(expr)
eq = sympy.Eq(expr, 0)
display(eq)
display(sympy.solve(eq)[0])
display(sympy.solve(eq)[1])

# 代数を使った式も取り扱える
expr = a * x**2 + b * x + c
display(expr)
eq = sympy.Eq(expr, 0)
display(eq)
display(sympy.solve(eq, x)[0])
display(sympy.solve(eq, x)[1])
```

```python
# 微分
expr = 2 * x**2 + 5 * x - 3
display(expr)
display(sympy.Derivative(expr, x))
display(sympy.Derivative(expr, x).doit())
display(sympy.diff(expr, x))
```

```python
# 一階微分、二階微分の等式
display(sympy.Eq(sympy.Derivative(expr, x), sympy.diff(expr, x)))
display(sympy.Eq(sympy.Derivative(expr, x, 2), sympy.diff(expr, x, 2)))
```

```python
# xについて微分して x=1を代入
expr = 2 * x**2 + 5 * x - 3
display(expr)
display(sympy.diff(expr))
sympy.diff(expr).subs(x, 1)
```

```python
# 偏微分表示(変数が2個)
expr = x**a
sympy.Eq(sympy.Derivative(expr, x), sympy.diff(expr, x))
```

```python
expr = sympy.sin(x)
sympy.Eq(sympy.Derivative(expr, x), sympy.diff(expr, x))
```

```python
expr = sympy.exp(x)
sympy.Eq(sympy.Derivative(expr, x), sympy.diff(expr, x))
```

```python
# 常微分方程式
f = sympy.Function("f")
eq = sympy.Eq(2 * f(x).diff(x, 1) + 5 * f(x), 0)
display(eq)
print("一般解")
display(sympy.dsolve(eq))
print("特殊解")
# 特殊解(境界条件)
ans = sympy.dsolve(eq, ics={f(0): 1})
display(ans)
# 特殊解 x=2 
display(ans.subs(x, 2))
display(sympy.Eq(f(2), sympy.exp(-5)))
plot(ans.rhs, (x, -10, 10))
```
