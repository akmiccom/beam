```python
from sympy import Derivative, Function
from sympy import symbols, integrate, diff, dsolve
from sympy import Eq
```

```python
I1, I2, P, E, L = symbols('I1, I2, P, E, L')
x = symbols('x')
v = Function('v')(x)
dvdx = Function('dvdx')(x)

phi = Eq(Derivative(dvdx, x), x/I1)
display(phi)
```

```python
dvdy_general_sol = dsolve(phi, dvdx)
dvdy_general_sol
```

```python
theta = Eq(Derivative(v, x), dvdy_general_sol.rhs)
theta
```

```python
v_general_sol = dsolve(theta, v)
v_general_sol
```

```python
display(phi)
display(theta)
display(v_general_sol)
```

```python
dvdy_special_sol = dsolve(phi, dvdx, ics={dvdx.subs(x, L): 0})
dvdy_special_sol = dsolve(phi, dvdx, ics={dvdx.subs(x, 0): 0})
dvdy_special_sol
```

## 断面二次モーメントが異なる場合の梁のたわみ計算
- 関数にして、連続的に計算できるようにする

```python
from sympy import symbols, Function, dsolve, Eq, Derivative, plot

P_value = 1000
E_value = 205000
I_value = 833.3

L_value = 20
n = [0, 10, 20]

L, P, E = symbols("L P E")
I = symbols("I")
x = symbols("x")
v = Function("v")(x)

M = P * (L - x)
deflect_angle = 0
deflect_value = 0

# 等式
eq = Eq(Derivative(v, x, x), M / E / I)
# 境界条件
ics = {v.subs(x, n[0]): deflect_value, Derivative(v, x).subs(x, n[0]): deflect_angle,}
# ソルバー
v_sol_ics = dsolve(eq, ics=ics)
# 値代入
v_numerical = v_sol_ics.rhs.subs({L: L_value, P: P_value, E: E_value, I: I_value})
# たわみ角・たわみ量出力
deflect_angle = Derivative(v_numerical).doit().subs(x, n[1])
deflect_value = v_numerical.subs(x, n[1])
print(f"{n[1]:.01f}mm たわみ角: {deflect_angle:.05f}")
print(f"{n[1]:.01f}mm たわみ量: {deflect_value:.05f}")

# for i in range(5, 21, 5):
#     angle = diff(v_numerical).subs(x, i)
#     value = v_numerical.subs(x, i)
#     print(f'{i:.01f}mm たわみ角: {angle:.05f}')
#     print(f'{i:.01f}mm たわみ量: {value:.05f}')
# plot(v_numerical, (x, 0, 20))
# 10.0mm たわみ角: 0.00088
# 10.0mm たわみ量: 0.00488
```

変数修正

```python
from sympy import symbols, Function, dsolve, Eq, Derivative, plot

P = 1000
L = 20
n = [0, 10, 20]

E_value = 205000
I_value = 833.3


def new_func(n, E_value, I_value):
    x = symbols("x")
    v = Function("v")(x)

    dd2vdx2 = P * (L - x) / E / I
    deflect_angle = 0
    deflect_value = 0

# 等式
    eq = Eq(Derivative(v, x, x), dd2vdx2)
# 境界条件
    ics = {v.subs(x, n[0]): deflect_value, Derivative(v, x).subs(x, n[0]): deflect_angle,}
# ソルバー
    v_sol_ics = dsolve(eq, ics=ics)
# 値代入
    v_numerical = v_sol_ics.rhs.subs({E: E_value, I: I_value})
# たわみ角・たわみ量出力
    deflect_angle = Derivative(v_numerical).doit().subs(x, n[1])
    deflect_value = v_numerical.subs(x, n[1])
    print(f"{n[1]:.01f}mm たわみ角: {deflect_angle:.05f}")
    print(f"{n[1]:.01f}mm たわみ量: {deflect_value:.05f}")

new_func(n, E_value, I_value)
```
