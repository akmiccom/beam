```python
from sympy import symbols, Function, dsolve, Eq
from sympy.abc import x

L, I1, I2 = symbols('L, I1, I2')
y = Function('y')(x)
dydx = Function('dydx')(x)
display(y)
display(dydx)
```

```python
eq2 = Eq(diff(y, x, x), x/I1)
display(eq2)
```

```python
eq1 = Eq(diff(dydx, x), x/I1)
display(eq1)
```

```python
dsolve(eq2, func=y)
```

```python
dsolve(eq1, func=dydx)
```

```python
dydx_sol = dsolve(eq1, func=dydx, ics={dydx.subs(x, L): 0})
dydx_sol
```

```python
y_sol = dsolve(eq2, func=y, ics={y.subs(x, L): 0, diff(y, x).subs(x, L): 0})
y_sol
```

```python
y_sol = dsolve(eq2, func=y, ics={y.subs(x, L/2): y.subs(x, L/2), diff(y, x).subs(x, L/2): diff(y, x).subs(x, L/2)})
y_sol
```
