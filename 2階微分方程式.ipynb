```python
from sympy import Function, dsolve, Eq, Derivative, symbols, pprint
```

```python
# 変数と未知関数の定義
x = symbols('x')
E, P = symbols('E P')
L, a = symbols('L a')
I1, I2 = symbols('I1 I2')
```

```python
y_1 = Function("y_1")(x)
dy_dx_1 = Function(f'dy_1/dx')(x)  # 一階微分：dy/dx
d2y_dx2_1 = Derivative(dy_dx_1)
display(y_1)
display(dy_dx_1)
display(d2y_dx2_1)
```

```python
# 二階微分方程式を一階微分方程式に変換
eq1 = Eq(d2y_dx2_1, x/I1)
display(eq1)
```

```python
# 一階微分方程式を解く
dy_dx_sol = dsolve(eq1, func=dy_dx_1, ics={dy_dx_1.subs(x, L): 0})
display(dy_dx_sol)
```

```python
# 一階微分方程式の解を使って元の二階微分方程式を再構築
eq2 = Eq(Derivative(y_1, x), dy_dx_sol.rhs)
display(eq2)
```

```python
# 二階微分方程式を解く
y_sol = dsolve(eq2, func=y_1, ics={y_1.subs(x, L): 0})
display(y_sol)
```

```python
from sympy import symbols, Function, Eq, Derivative, dsolve, solve

# 変数と関数の定義
x = symbols('x')
y1 = Function('y1')(x)
y2 = Function('y2')(x)
P, E, I1, I2, L = symbols('P E I1 I2 L')

# 部位1の微分方程式
eq1 = Eq(Derivative(y1, x), x/I1)  # 0 <= x < L/2
display(eq1)

# 部位2の微分方程式
eq2 = Eq(Derivative(y2, x), x/I2)  # L/2 <= x <= L
display(eq2)

# 部位1の解を求める
y1_sol = dsolve(eq1, y1, ics={y1.subs(x, L): 0})
display(y1_sol)

# 部位2の解を求める
y2_sol = dsolve(eq2, y2)
display(y2_sol)

# 部位1の解を取り出す
y1 = y1_sol.rhs
display(y1)  
# 部位2の解を取り出す
y2 = y2_sol.rhs
display(y2)

# dsolve(eq, y1, ics={y2.subs(x, L/2): y1.subs(x, L/2)})

# 境界条件 たわみの連続性: y1(L/2) = y2(L/2)
continuity_eq = Eq(y1.subs(x, L/2), y2.subs(x, L/2))
display(continuity_eq)

# 傾きの連続性: dy1/dx(L/2) = dy2/dx(L/2)
# slope_eq = Eq(Derivative(y1, x).subs(x, L/2), Derivative(y2, x).subs(x, L/2))

# 連立方程式として定数を解く
constants = solve([continuity_eq])
display(constants[0])

continuity_eq.subs(constants[0])
```

```python
from sympy import symbols, Function, Eq, Derivative, dsolve, solve

# 変数と関数の定義
x = symbols('x')
y1 = Function('y1')(x)
y2 = Function('y2')(x)
P, E, I1, I2, L = symbols('P E I1 I2 L')

# 部位1の微分方程式
eq3 = Eq(Derivative(y1, x, 2), x/I1)  # 0 <= x < L/2
y3_sol = dsolve(eq3, y1, ics={y1.subs(x, L): 0, Derivative(y1, x).subs(x, L): 0})
display(y3_sol)


# # 部位2の微分方程式
eq4 = Eq(Derivative(y2, x, 2), x/I1)  # 0 <= x < L/2
y4_sol = dsolve(eq4, y2)
display(y4_sol)
```

```python
from sympy import symbols, Function, Eq, Derivative, dsolve

# 変数と関数の定義
x = symbols('x')
y1 = Function('y1')(x)
y2 = Function('y2')(x)
P, E, I1, I2, L = symbols('P E I1 I2 L')

# 部位1の微分方程式
eq1 = Eq(Derivative(y1, x), P/(E * I1))  # 0 <= x < L/2

# 部位2の微分方程式
eq2 = Eq(Derivative(y2, x), P/(E * I2))  # L/2 <= x <= L

# 境界条件の設定
# 部位1: y1(0) = 0, dy1/dx(0) = 0
# 部位2: y2(L) = 0, dy2/dx(L) = 0
boundary_conditions_1 = {y1.subs(x, 0): 0, Derivative(y1, x).subs(x, 0): 0}
boundary_conditions_2 = {y2.subs(x, L): 0, Derivative(y2, x).subs(x, L): 0}

# 部位1の解を求める（境界条件付き）
y1_sol = dsolve(eq1, y1, ics=boundary_conditions_1)

# 部位2の解を求める（境界条件付き）
y2_sol = dsolve(eq2, y2, ics=boundary_conditions_2)

# 解を表示
display(y1_sol)
display(y2_sol)

# 連続性の境界条件を手動で設定
# y1(L/2) = y2(L/2) と dy1/dx(L/2) = dy2/dx(L/2)
y1 = y1_sol.rhs
y2 = y2_sol.rhs
continuity_eq = Eq(y1.subs(x, L/2), y2.subs(x, L/2))
# slope_eq = Eq(Derivative(y1, x).subs(x, L/2), Derivative(y2, x).subs(x, L/2))

# 定数を解く
constants = solve([continuity_eq])

# 定数を代入して最終的な解を取得
y1_final = y1.subs(constants)
y2_final = y2.subs(constants)

# 最終的な解を表示
display(y1_final)
display(y2_final)

```
